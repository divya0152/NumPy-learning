Lesson 3 — NumPy Indexing
-------------------------------------------------
Indexing means accessing a particular element from an array.
Python and NumPy use zero-based indexing.
That means counting starts from 0, not 1.

```python

import numpy as np
numbers=np.array([10,20,30,40,50])
print(numbers[3])
print(numbers[-2])
```

Negative indexing
---------------------
You can also count from the end:
Value:     10   20   30   40   50
Index:      0    1    2    3    4
Negative:  -5   -4   -3   -2   -1

```python
import numpy as np
marks = np.array([
    [80, 90, 70],
    [75, 85, 95],
    [88, 92, 84]
])
# to access an elemnt 
#marks[row,column]
print(marks[0,1])
print(marks[:,0]) #first column(All rows, column 0)
print(marks[:,2]) #third column
print(marks[:,-1] )#third column
print(marks[2])    #last row
print(marks[:, :]) # all rows , columns
```

Lesson 4 — NumPy Slicing
---------------------------------------------
Slicing means taking a portion of an array.
The basic syntax is:
array[start:stop]
⚠️ The stop index is not included.

```python
import numpy as np
numbers=np.array([10,20,30,40,50])
print(numbers[1:4])#Start at index 1, stop before index 4.
print(numbers[0:3])#Start at index 0, stop before index 3.
print(numbers[:3]) #start from beginning
print(numbers[2:]) #continue until the end
print(numbers[:])# entire array
```

2D Slicing
-----------------------------------------
marks[row_start:row_stop, column_start:column_stop]

```python
import numpy as np
marks = np.array([
    [80, 90, 70, 60],
    [75, 85, 95, 65],
    [88, 92, 84, 72]
])
print(marks[0:2,1:3])
```

STEP:
---------------------------------
Use the step value to determine the step of the slicing:

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])
print(arr[::2])
```

Lesson 9 — NumPy Aggregation Functions
--------------------------------------
Aggregation functions summarize an entire dataset into a single value.

```python
import numpy as np
marks = np.array([80, 90, 70, 60, 100])
sum=np.sum(marks)
print("Sum:",sum)
mean=np.mean(marks)
print("Mean:",mean)
median=np.median(marks)
print("Median:",median)
max=np.max(marks)
print("Max",max)
min=np.min(marks)
print("MIN",min)
print(np.std(marks))
print(np.var(marks))
```

Important Concept:axis
--------------------------

axis=0 → column-wise → result for each column


axis=1 → row-wise → result for each row

```python
marks = np.array([
    [80, 90, 70],
    [75, 85, 95],
    [88, 92, 84]
])
Total_sum=np.sum(marks)
print("Total_sum:",Total_sum)
Max_value=np.max(marks)
print("Max_value:",Max_value)
Avg_SubjectWise=np.mean(marks,axis=0)
print("Avg_SubjectWise:",Avg_SubjectWise)
Total_marksStudentwise=np.sum(marks,axis=1)
print("Total_marksStudentwise:",Total_marksStudentwise)
print(np.min(marks,axis=0))
```

Lesson 10 — Comparison Operations & Boolean Indexing
----------------------------------------------------

array[condition] → returns only the elements where the condition is True.


```python
marks = np.array([45, 67, 82, 91, 55])
marks > 80
marks[marks > 60]
```

```python
marks = np.array([45, 67, 82, 91, 55, 73, 88]) 
print(marks>70) 
print(marks[marks>70]) 
print(marks[marks<60])
print(marks[marks>=80]) 
print(marks[(marks >= 60) & (marks <= 80)])
```

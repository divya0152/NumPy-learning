---------------------------------------------
Slicing means taking a portion of an array.
The basic syntax is:

⚠️ The stop index is not included.

```python
import numpy as np
numbers=np.array([10,20,30,40,50])
print(numbers[1:4])#Start at index 1, stop before index 4.
print(numbers[0:3])#Start at index 0, stop before index 3.
print(numbers[:3]) #start from beginning
print(numbers[2:]) #continue until the end
print(numbers[:])# entire array
```

Lesson 14 — flatten() vs ravel()
---------------------------------
Both are used to convert a multidimensional array into 1D
<--flatten() always returns a copy of the array, while ravel() returns a flattened view when possible, making ravel() generally more memory-efficient."-->

```python
import numpy as np
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
arr.shape
flat = arr.flatten()
print(flat)
flat[0]=100
print("The original generally remains:flatten()made a copy")
print(arr)
flats = arr.ravel()
print(flats)
flats[0] = 100
print("You may see:because the raveled result can share memory with arr")
print(arr)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print("Use flatten() and print the result")
flat=arr.flatten()
print(flat)
print(flat.shape)
flat[0]=999
print(arr)
print("Use ravel() and print the result")
flats=arr.ravel()
print(flats)
print(flats.shape)
flats[0]=999
print(arr)
```

Topic 3: Memory Efficiency — view() vs copy()

A view creates a new array object that shares the same underlying data, while a copy creates an independent array with its own data.

| Method   | Shares memory? | Changing it affects original? |
| -------- | -------------- | ----------------------------- |
| `view()` | ✅ Yes          | ✅ Usually yes                 |
| `copy()` | ❌ No           | ❌ No                          |


```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
view_arr = arr.view()
view_arr[0]=999
print("view_arr:",view_arr)
print("arr",arr)
```

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
copy_arr = arr.copy()
copy_arr[0]=999
print(copy_arr)
print(arr)
```

3. np.shares_memory()

NumPy provides:
np.shares_memory(array1, array2)

It returns:

True → both arrays share the same memory
False → they don't share memory

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

view_arr = arr.view()
copy_arr = arr.copy()

print(np.shares_memory(arr, view_arr))
print(np.shares_memory(arr, copy_arr))
```

What is Broadcasting?

Broadcasting allows NumPy to perform operations between arrays of different shapes.
When NumPy compares shapes, it works from the rightmost dimension.

🧠 Simple rule

Two dimensions are compatible if:

They are equal, or
One of them is 1, or
One dimension doesn't exist because the smaller array has fewer dimensions.

Topic 5: Memory & Array Efficiency

1. itemsize — memory used by one element

2. nbytes — total memory used by array data

   There are 4 elements × 8 bytes:
    4 × 8 = 32 bytes
   

🧠 Important interview point

itemsize tells us the memory occupied by one array element, while nbytes tells us the total memory occupied by the array's elements.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50], dtype=np.int32)

print("dtype:", arr.dtype)
print("itemsize:", arr.itemsize)
print("nbytes:", arr.nbytes)
print("size:", arr.size)
```

```python
arr64 = np.array([1, 2, 3, 4, 5], dtype=np.int64)
arr32 = np.array([1, 2, 3, 4, 5], dtype=np.int32)

print(arr64.itemsize)
print(arr64.nbytes)

print(arr32.itemsize)
print(arr32.nbytes)
```

Topic 8 — np.einsum() & Advanced Numerical Operations

1. What is np.einsum()?

einsum stands for Einstein summation notation.

It lets us perform operations such as:

Sum ,
Dot product ,
Matrix multiplication ,
Transpose ,
Element-wise multiplication + summation .

```python
import numpy as np
a = np.array([2, 4, 6])
b = np.array([3, 5, 7])

result = np.einsum('i,i->', a, b)

print(result)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
np.einsum('ik,kj->ij', A, B)
```

🧠 einsum() patterns to remember
----------------------------------
| Operation             | `einsum()`    |
| --------------------- | ------------- |
| Dot product           | `'i,i->'`     |
| Matrix multiplication | `'ik,kj->ij'` |

```python
import numpy as np

students = np.array([
    [101, 78, 85, 92],
    [102, 45, 67, 55],
    [103, 88, 91, 84],
    [104, 32, 49, 40],
    [105, 72, 65, 78],
    [106, 95, 89, 93]
])

# 1. Calculate overall average for each student
student_averages = np.mean(students[:, 1:], axis=1)

print("Averages:", student_averages)

# 2. Classify students
result = np.where(
    student_averages >= 85,
    "Excellent",
    np.where(
        student_averages >= 70,
        "Good",
        np.where(
            student_averages >= 50,
            "Average",
            "Fail"
        )
    )
)

print("Result:", result)
```

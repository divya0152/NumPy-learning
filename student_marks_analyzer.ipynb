# Student Marks Analyzer

## NumPy Mini Project

This mini project applies the NumPy concepts learned in `NumPy_Learning.ipynb` to a small student-performance dataset.

### Objectives
- Analyze student marks
- Calculate subject-wise and student-wise statistics
- Filter students using Boolean indexing
- Use `np.where()` for conditional transformations
- Find the highest-performing student
- Practice NumPy indexing, slicing, aggregation, and vectorization

## 1. Import NumPy and Create the Dataset

**Columns:**

`Student_ID | Python | SQL | NumPy`

```python
import numpy as np

students = np.array([
    [101, 78, 85, 92],
    [102, 45, 67, 55],
    [103, 88, 91, 84],
    [104, 32, 49, 40],
    [105, 72, 65, 78],
    [106, 95, 89, 93]
])

print(students)
```

## 2. Dataset Properties

Find the shape, total number of elements, and number of dimensions.

```python
print("Shape:", students.shape)
print("Size:", students.size)
print("Dimensions:", students.ndim)
```

## 3. Extract Subject Marks

Extract the Python, SQL, and NumPy marks. The Student ID column should not be included.

```python
python_marks = students[:, 1]
sql_marks = students[:, 2]
numpy_marks = students[:, 3]

print("Python:", python_marks)
print("SQL:", sql_marks)
print("NumPy:", numpy_marks)
```

## 4. Subject-wise Average

Calculate the average mark for Python, SQL, and NumPy.

```python
subject_averages = np.mean(students[:, 1:], axis=0)

print("Python average:", subject_averages[0])
print("SQL average:", subject_averages[1])
print("NumPy average:", subject_averages[2])
```

## 5. Student-wise Average

Calculate the overall average for every student.

```python
student_averages = np.mean(students[:, 1:], axis=1)

print("Student averages:", student_averages)
```

## 6. Student with the Highest Overall Average

Use `np.argmax()` to find the index and then use that index to retrieve the Student ID.

```python
highest_index = np.argmax(student_averages)
highest_student_id = students[highest_index, 0]

print("Highest average:", student_averages[highest_index])
print("Student ID:", highest_student_id)
```

## 7. Students with Average Below 60

Use Boolean indexing to find students whose overall average is below 60.

```python
below_60 = student_averages < 60

print("Averages below 60:", student_averages[below_60])
print("Student IDs:", students[below_60, 0])
```

## 8. Pass / Fail Classification

A student passes when their overall average is at least 50.

```python
result = np.where(student_averages >= 50, "Pass", "Fail")

print(result)
```

## 9. Add a 5-Mark Bonus

For subject marks below 50, add a fixed 5-mark bonus.

Important: the Student ID must not be modified.

```python
subject_marks = students[:, 1:]

bonus = np.where(subject_marks < 50, 5, 0)
final_marks = subject_marks + bonus

print("Bonus:")
print(bonus)

print("Final subject marks:")
print(final_marks)
```

### Final Dataset with Student IDs

```python
final_students = np.column_stack((students[:, 0], final_marks))

print(final_students)
```

## 10. Final Analysis

Find:
1. Highest Python mark
2. Lowest SQL mark
3. Average NumPy mark
4. Number of students with average >= 70
5. Student ID with the highest overall average

```python
print("Highest Python mark:", np.max(students[:, 1]))
print("Lowest SQL mark:", np.min(students[:, 2]))
print("Average NumPy mark:", np.mean(students[:, 3]))

count_70 = np.sum(student_averages >= 70)
print("Students with average >= 70:", count_70)

highest_index = np.argmax(student_averages)
print("Student ID with highest average:", students[highest_index, 0])
```

## 11. Performance Classification

Classify students based on their overall average:

- `>= 85` → Excellent
- `>= 70` → Good
- `>= 50` → Average
- `< 50` → Fail

```python
performance = np.where(
    student_averages >= 85,
    "Excellent",
    np.where(
        student_averages >= 70,
        "Good",
        np.where(
            student_averages >= 50,
            "Average",
            "Fail"
        )
    )
)

print(performance)
```

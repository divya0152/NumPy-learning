# 💼 Salary Analysis using NumPy
This project analyzes employee salary, experience, and performance data using NumPy.

The goal is to apply NumPy concepts to a practical, internship-oriented dataset.


## 🎯 Objectives

- Analyze employee salaries
- Analyze employee experience
- Analyze employee performance
- Use NumPy statistical functions
- Practice Boolean indexing and `np.where()`
- Apply conditional salary increases
- Identify employees based on multiple conditions

```python
import numpy as np 
#Employee_ID | Salary | Experience | Performance
employees=np.array([
    [101,25000,2,65],
    [102,45000,5,82],
    [103,32000,3,70],
    [104,60000,8,95],
    [105,28000,1,55],
    [106,52000,6,88]
])
print(employees)
```

```python
print("Shape",employees.shape)
print("Size",employees.size)
print("ndim:",employees.ndim)
```

```python
print("Employees IDs:",employees[:,0])
print("Salaries:",employees[:,1])
print("Experience:",employees[:,2])
print("Performance:",employees[:,3])
```

## 💰 Salary Analysis

Let's calculate the average, maximum, minimum, and standard deviation of employee salaries.

```python
print("Average Salary:",np.mean(employees[:,1],axis=0))
print("Maximum Salary:",np.max(employees[:,1],axis=0))
print("Minimum Salary:",np.min(employees[:,1],axis=0))
print("Standard Deviation Of Salary:",np.std(employees[:,1],axis=0))
```

```python
high_performance_index=np.where(employees[:,3]>80)[0]
high_performance_ids=employees[high_performance_index,0]
print("High Performance Employee IDs:", high_performance_ids)
```

```python
salaries = employees[:, 1]
print("salaries:",salaries )
salary_increase=np.where(salaries<40000,salaries*0.1,0)
final_salary=salaries+salary_increase
print("salary_increase:",salary_increase)
print("final_salary:",final_salary)
```

```python
avg_performance=np.mean(employees[:,3])
print(avg_performance)
high_performance_index=np.where(employees[:,3]>=80)[0]
high_performance=employees[high_performance_index,0]
print("High Performance Employee IDs:", high_performance)
```

```python
performance = employees[:, 3]

highest_performance_index = np.argmax(performance)

highest_performance_id = employees[highest_performance_index, 0]

print("Highest Performance:", performance[highest_performance_index])
print("Employee ID:", highest_performance_id)
```

## 📈 Experience Analysis

Let's analyze employee experience and identify the most experienced employees.

```python
experience=employees[:,2]
print("experience:",experience)
avg_experience=np.mean(experience)
print("avg_experience:",avg_experience)
highest_experience_index=np.argmax(experience)
highest_experience_id=employees[highest_experience_index,0]
print("Highest Experience:", experience[highest_experience_index])
print("Employee ID:", highest_experience_id)
high_experience_index = np.where(experience > 4)[0]
high_experience_id=employees[high_experience_index ,0]
print("Employee IDs having more than 4 years of experience.:",high_experience_id)
```

```python
performance_category=np.where(performance>=90,"Excellent",np.where(performance>=80,"GOOD",np.where(performance>=60,"AVERAGE","POOR")))
print("RESULT:",performance_category)
```

```python
salaries=employees[:,1]
print(salaries)
highest_salary=np.max(salaries)
print("Highest salary:",highest_salary)
highest_paid_index=np.argmax(salaries)
print("Employee ID of the highest-paid employee:",employees[highest_paid_index,0])
lowest_salary=np.min(salaries)
print("Lowest salary:",lowest_salary)
lowest_paid_index=np.argmin(salaries)
print("Employee ID of the lowest-paid employee:",employees[lowest_paid_index,0])
count40=np.sum(salaries > 40000)
print(count40)
```
